```python
# 13.1 Write the current date as a string to the text file today.txt
import datetime
today_date = datetime.date.today()
date_string = today_date.strftime("%Y-%m-%d")
with open("today.txt","w") as file:
    file.write(date_string)
```

```python
# 13.2 Read the text file today.txt into the string today_string
file = open("today.txt","r")
today_string = file.read()
file.close()
```

```python
# 13.3 Parse the date from today_string
import time
fmt = "%Y-%m-%d"
time.strptime("2025-10-01", fmt)
```

```python
# 15.1 Use multiprocessing to create three separate processes
# Make each one wait a random number of seconds between zero and one,
# print the current time, and then exit.
import multiprocessing
import datetime
import random
import time

def time_current():
    wait = random.uniform(0, 1)
    time.sleep(wait)
    
    today_string = datetime.datetime.now()
    time_string = today_string.strftime("%I:%M:%S %p")
    print(time_string)

if __name__ == "__main__":
    processes = []
    for i in range(3):
        p = multiprocessing.Process(target=time_current)
        p.start()
        p.join()
```
